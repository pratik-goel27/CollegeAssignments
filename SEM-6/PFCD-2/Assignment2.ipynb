```python
# Q1 :-
def power(base, exponent):
    if exponent == 1:
        return base
    return base * power(base, exponent - 1)
    
print(power(3, 4))
```

```python
# Q3 :-
def generate_increasing(n, current="", last_digit=0):
    if len(current) == n:
        print(current, end=" ")
        return
    
    for digit in range(last_digit + 1, 10):
        generate_increasing(n, current + str(digit), digit)

print(generate_increasing(2, "", 0))
```

```python
# Q4 :-
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

def fibonacci_iterative(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

print(fibonacci(6))
print(fibonacci_iterative(6))
```
